# New Topic: Hypothesis Testing (Learning Notes)

Hypothesis testing helps determine whether an observed difference is statistically significant.

### Steps
1. Define the business question.
2. State hypotheses.
3. Choose a significance level (α = 0.05).
4. Select the appropriate statistical test.
5. Calculate the p-value.
6. Draw a conclusion.

### Terms
- **Null hypothesis (H₀):** No difference or relationship.
- **Alternative hypothesis (H₁):** A difference or relationship exists.
- **p-value:** Probability of observing the data if H₀ is true.
- If p < 0.05 → Reject H₀.
- If p ≥ 0.05 → Fail to reject H₀.

### Tests you'll use
- Independent t-test → Compare two group means.
- Pearson correlation → Relationship between two numeric variables.
- Chi-square test → Relationship between categorical variables.


```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_excel('Online Retail.xlsx')
df.head()
```

```
# Basic information
display(df.head())
display(df.tail())
print(df.shape)
df.info()
```

```
# Missing values
df.isna().sum().sort_values(ascending=False)
```

```
# Remove cancelled invoices and invalid rows
df = df[~df['InvoiceNo'].astype(str).str.startswith('C')]
df = df[df['Quantity'] > 0]
df = df[df['UnitPrice'] > 0]
df = df.drop_duplicates()
df['CustomerID'] = df['CustomerID'].astype('Int64')
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
print(df.shape)
```

```
# Feature Engineering
df['Revenue'] = df['Quantity'] * df['UnitPrice']
df['Month'] = df['InvoiceDate'].dt.month_name()
df['Weekday'] = df['InvoiceDate'].dt.day_name()
df['Hour'] = df['InvoiceDate'].dt.hour
df.head()
```

# Exploratory Data Analysis

Suggested analyses:

```
# Top 10 countries by revenue
country_rev = df.groupby('Country')['Revenue'].sum().sort_values(ascending=False).head(10)
country_rev.plot(kind='bar', figsize=(8,4))
plt.ylabel('Revenue')
plt.tight_layout()
plt.show()
```

```
# Summary statistics
display(df[['Quantity','UnitPrice','Revenue']].describe())
print(df[['Quantity','UnitPrice','Revenue']].corr())
```

# Hypothesis Test Example

## Business Question
Do UK customers spend more per transaction than customers from other countries?

H₀: Mean transaction revenue is the same.

H₁: Mean transaction revenue is different.


```
from scipy.stats import ttest_ind

uk = df[df['Country']=='United Kingdom']['Revenue']
nonuk = df[df['Country']!='United Kingdom']['Revenue']

stat, p = ttest_ind(uk, nonuk, equal_var=False)

print('t-statistic:', stat)
print('p-value:', p)

if p < 0.05:
    print('Reject H0')
else:
    print('Fail to reject H0')
```
