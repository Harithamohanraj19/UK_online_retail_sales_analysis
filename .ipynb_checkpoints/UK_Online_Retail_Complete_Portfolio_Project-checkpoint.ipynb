## Executive Summary

This project analyses transaction data from a UK-based non-store online retailer that mainly sells all-occasion gifts. The objective is to identify the products, customers, countries and time periods that contribute most to revenue, and to translate the findings into practical business recommendations.

The analysis covers:

- data quality assessment and cleaning;
- sales, product, country, customer and time-based analysis;
- order-level feature engineering;
- descriptive statistics and correlation;
- statistically valid hypothesis tests;
- business recommendations and project limitations.

> **Important analytical decision:** Product-line records are useful for product analysis, but customer spending comparisons should be performed at **invoice/order level**. Treating every product line as an independent customer purchase would overstate the sample size.

## 1. Business Problem

The retailer wants to make better decisions about product promotion, international growth, inventory planning and customer retention.

### Main analytical question

**How can the retailer use historical transaction data to improve sales performance and customer engagement?**

### Supporting questions

1. How much valid revenue was generated during the period?
2. Which products generated the most quantity and revenue?
3. Which countries contributed the most revenue outside the UK?
4. How did revenue change by month, weekday and hour?
5. Which customers generated the most value?
6. How concentrated was revenue among customers?
7. Is average order value different between UK and non-UK customers?
8. Is average order value different between weekdays and weekends?

## 2. Dataset Description

Each row represents one product line within an invoice.

| Column | Meaning |
|---|---|
| `InvoiceNo` | Invoice number; values beginning with `C` represent cancellations |
| `StockCode` | Product code |
| `Description` | Product description |
| `Quantity` | Number of units on the invoice line |
| `InvoiceDate` | Date and time of the transaction |
| `UnitPrice` | Price per unit in pounds sterling |
| `CustomerID` | Unique customer identifier |
| `Country` | Customer country |

The raw dataset contains both completed purchases and records that require careful interpretation, including cancellations, returns, missing customer identifiers, duplicates, zero prices and unusual stock codes.

## 6. Data Cleaning Strategy

### Quick revision

Data cleaning means deciding which records are valid for the specific business question. It does not mean deleting every unusual value automatically.

### Decisions used in the main sales analysis

- Remove exact duplicate rows.
- Remove cancelled invoices (`InvoiceNo` begins with `C`).
- Keep only positive quantities and positive unit prices.
- Keep rows with missing `CustomerID` for overall product and sales analysis.
- Create a separate customer dataset that excludes missing `CustomerID`.
- Retain unusually large valid orders; do not remove them merely because they are statistical outliers.

A separate return/cancellation analysis is outside this project's main scope, but the excluded records are counted and documented.

````python
df = df_raw.copy()

cleaning_log = []

def log_step(step, before, after):
    cleaning_log.append({
        "Step": step,
        "Rows before": before,
        "Rows after": after,
        "Rows removed": before - after
    })

before = len(df)
df = df.drop_duplicates().copy()
log_step("Remove exact duplicate rows", before, len(df))

before = len(df)
df = df[~df["InvoiceNo"].astype(str).str.startswith("C")].copy()
log_step("Remove cancelled invoices", before, len(df))

before = len(df)
df = df[df["Quantity"] > 0].copy()
log_step("Keep positive quantities", before, len(df))

before = len(df)
df = df[df["UnitPrice"] > 0].copy()
log_step("Keep positive unit prices", before, len(df))

df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
df["InvoiceNo"] = df["InvoiceNo"].astype(str)

cleaning_log = pd.DataFrame(cleaning_log)
cleaning_log
````

````python
print(f"Clean sales rows: {len(df):,}")
print(f"Rows retained: {len(df) / len(df_raw):.1%}")
print(f"Missing CustomerID values retained for non-customer analysis: {df['CustomerID'].isna().sum():,}")
````

## 7. Feature Engineering

### Quick revision

Feature engineering creates variables that make the business questions easier to answer.

The following features are created:

- `Revenue` = `Quantity × UnitPrice`
- year, month, weekday, date and hour
- UK versus non-UK market group
- invoice-level order value, quantity and number of product lines

````python
df["Revenue"] = df["Quantity"] * df["UnitPrice"]
df["Year"] = df["InvoiceDate"].dt.year
df["MonthPeriod"] = df["InvoiceDate"].dt.to_period("M")
df["Month"] = df["InvoiceDate"].dt.month_name()
df["Weekday"] = df["InvoiceDate"].dt.day_name()
df["WeekdayNumber"] = df["InvoiceDate"].dt.dayofweek
df["DayType"] = np.where(df["WeekdayNumber"] < 5, "Weekday", "Weekend")
df["Hour"] = df["InvoiceDate"].dt.hour
df["Date"] = df["InvoiceDate"].dt.date
df["Market"] = np.where(df["Country"] == "United Kingdom", "United Kingdom", "Non-UK")

order_summary = (
    df.groupby(["InvoiceNo", "InvoiceDate", "Country", "Market", "DayType"], as_index=False)
      .agg(
          OrderValue=("Revenue", "sum"),
          Units=("Quantity", "sum"),
          ProductLines=("StockCode", "count"),
          UniqueProducts=("StockCode", "nunique")
      )
)

order_summary["MonthPeriod"] = order_summary["InvoiceDate"].dt.to_period("M")
order_summary["Weekday"] = order_summary["InvoiceDate"].dt.day_name()
order_summary["WeekdayNumber"] = order_summary["InvoiceDate"].dt.dayofweek
order_summary["Hour"] = order_summary["InvoiceDate"].dt.hour

customer_lines = df.dropna(subset=["CustomerID"]).copy()
customer_lines["CustomerID"] = customer_lines["CustomerID"].astype(int)

df.head()
````

## 15. Customer Value Analysis

Customer analysis excludes rows without `CustomerID` because individual customers cannot be tracked reliably without an identifier.

````python
customer_summary = (
    customer_lines.groupby("CustomerID", as_index=False)
      .agg(
          Revenue=("Revenue", "sum"),
          Orders=("InvoiceNo", "nunique"),
          Units=("Quantity", "sum"),
          LastPurchase=("InvoiceDate", "max"),
          FirstPurchase=("InvoiceDate", "min")
      )
)

customer_summary["AverageOrderValue"] = customer_summary["Revenue"] / customer_summary["Orders"]
customer_summary["CustomerLifetimeDays"] = (
    customer_summary["LastPurchase"] - customer_summary["FirstPurchase"]
).dt.days

top_customers = customer_summary.nlargest(10, "Revenue").sort_values("Revenue")

fig, ax = plt.subplots(figsize=(9, 5))
ax.barh(top_customers["CustomerID"].astype(str), top_customers["Revenue"])
ax.set_title("Top 10 Customers by Revenue")
ax.set_xlabel("Revenue (£)")
ax.set_ylabel("Customer ID")
plt.tight_layout()
plt.show()

customer_summary.nlargest(10, "Revenue")
````

# 17. Hypothesis Testing

Hypothesis testing is used here to decide whether observed differences are likely to reflect genuine population differences rather than random sampling variation.

A significance level of **α = 0.05** is used.

Because order values are highly skewed and group variances are not assumed to be equal, the project reports:

1. **Welch's independent-samples t-test** on order value;
2. **Mann–Whitney U test** as a non-parametric robustness check;
3. effect size and group medians to assess practical importance.

Statistical significance alone does not prove that a difference is commercially important.

## 17.1 Test One: UK vs Non-UK Order Value

### Business question

Is the average completed order value different between UK and non-UK orders?

### Hypotheses

- **H₀:** Mean UK order value equals mean non-UK order value.
- **H₁:** Mean UK order value differs from mean non-UK order value.

````python
uk_orders = order_summary.loc[
    order_summary["Market"] == "United Kingdom", "OrderValue"
]
nonuk_orders = order_summary.loc[
    order_summary["Market"] == "Non-UK", "OrderValue"
]

welch_uk = stats.ttest_ind(
    uk_orders,
    nonuk_orders,
    equal_var=False,
    nan_policy="omit"
)

mannwhitney_uk = stats.mannwhitneyu(
    uk_orders,
    nonuk_orders,
    alternative="two-sided"
)

def cohens_d_independent(a, b):
    a = np.asarray(a, dtype=float)
    b = np.asarray(b, dtype=float)
    pooled_sd = np.sqrt(
        ((len(a) - 1) * np.var(a, ddof=1) + (len(b) - 1) * np.var(b, ddof=1))
        / (len(a) + len(b) - 2)
    )
    return (np.mean(a) - np.mean(b)) / pooled_sd

uk_effect = cohens_d_independent(uk_orders, nonuk_orders)

uk_test_results = pd.DataFrame({
    "Group": ["United Kingdom", "Non-UK"],
    "Orders": [len(uk_orders), len(nonuk_orders)],
    "Mean order value": [uk_orders.mean(), nonuk_orders.mean()],
    "Median order value": [uk_orders.median(), nonuk_orders.median()],
    "Standard deviation": [uk_orders.std(), nonuk_orders.std()]
})

display(uk_test_results)
print(f"Welch t-statistic: {welch_uk.statistic:.4f}")
print(f"Welch p-value: {welch_uk.pvalue:.6g}")
print(f"Mann–Whitney p-value: {mannwhitney_uk.pvalue:.6g}")
print(f"Cohen's d: {uk_effect:.4f}")

if welch_uk.pvalue < 0.05:
    print("Decision: Reject H₀. Mean order value differs significantly between the groups.")
else:
    print("Decision: Fail to reject H₀. The evidence is insufficient to show a difference.")
````

````python
plot_data = [
    np.log1p(uk_orders),
    np.log1p(nonuk_orders)
]

fig, ax = plt.subplots(figsize=(7, 5))
ax.boxplot(plot_data, labels=["United Kingdom", "Non-UK"], showfliers=False)
ax.set_title("Log-Transformed Order Value by Market")
ax.set_ylabel("log(1 + order value)")
plt.tight_layout()
plt.show()
````

## 17.2 Test Two: Weekday vs Weekend Order Value

### Business question

Is average order value different between weekday and weekend orders?

### Hypotheses

- **H₀:** Mean weekday order value equals mean weekend order value.
- **H₁:** Mean weekday order value differs from mean weekend order value.

> The dataset contains no Saturday transactions, so the weekend group is mainly Sunday. This limitation must be considered when interpreting the result.

````python
weekday_orders = order_summary.loc[
    order_summary["DayType"] == "Weekday", "OrderValue"
]
weekend_orders = order_summary.loc[
    order_summary["DayType"] == "Weekend", "OrderValue"
]

welch_day = stats.ttest_ind(
    weekday_orders,
    weekend_orders,
    equal_var=False,
    nan_policy="omit"
)

mannwhitney_day = stats.mannwhitneyu(
    weekday_orders,
    weekend_orders,
    alternative="two-sided"
)

day_effect = cohens_d_independent(weekday_orders, weekend_orders)

day_test_results = pd.DataFrame({
    "Group": ["Weekday", "Weekend"],
    "Orders": [len(weekday_orders), len(weekend_orders)],
    "Mean order value": [weekday_orders.mean(), weekend_orders.mean()],
    "Median order value": [weekday_orders.median(), weekend_orders.median()],
    "Standard deviation": [weekday_orders.std(), weekend_orders.std()]
})

display(day_test_results)
print(f"Welch t-statistic: {welch_day.statistic:.4f}")
print(f"Welch p-value: {welch_day.pvalue:.6g}")
print(f"Mann–Whitney p-value: {mannwhitney_day.pvalue:.6g}")
print(f"Cohen's d: {day_effect:.4f}")

if welch_day.pvalue < 0.05:
    print("Decision: Reject H₀. Mean order value differs significantly by day type.")
else:
    print("Decision: Fail to reject H₀. The evidence is insufficient to show a difference.")
````

## 19. Business Recommendations

### 1. Protect and develop high-value customers

Revenue is concentrated among a relatively small group of customers. The company should introduce targeted retention activity such as account management, loyalty benefits, early access and personalised product recommendations.

### 2. Plan inventory around seasonal demand

Revenue rises sharply during the strongest months. Best-selling and high-revenue products should be stocked earlier, and supplier lead times should be reviewed before the seasonal peak.

### 3. Separate volume products from value products

The products with the highest unit sales are not necessarily the products with the highest revenue. Product decisions should therefore use several KPIs: units, revenue, order frequency and margin where available.

### 4. Reduce dependence on the domestic market

The UK provides the majority of revenue. Strong international markets should be investigated for targeted campaigns, localised promotions and distribution improvements.

### 5. Use order-level metrics for customer decisions

Customer and market comparisons should be based on complete invoices rather than individual product lines. This avoids overstating the number of independent purchases.

### 6. Validate statistical findings commercially

A statistically significant result may have a small effect size. Before changing pricing or campaign strategy, the company should consider the size of the difference, implementation cost and expected return.

## 20. Limitations

- The dataset covers only one retailer and slightly more than one year.
- December 2011 is incomplete and contains only the first nine days.
- Product cost, profit margin, marketing spend and customer demographics are unavailable.
- Many rows have no `CustomerID`, limiting customer-level analysis.
- Cancelled and return transactions were excluded from the main sales analysis rather than modelled separately.
- The dataset contains no Saturday sales and weekend comparisons are therefore limited.
- Statistical association does not establish causation.
- Large wholesale purchases create highly skewed order values.

## 21. Conclusion

This analysis converted raw transaction records into product-line, order-level and customer-level views. The results identify the main sources of revenue, reveal seasonal and market concentration, rank products and customers, and test whether important customer groups differ statistically.

The strongest business opportunities are to retain high-value customers, improve seasonal inventory planning, support proven international markets and evaluate products using both volume and revenue measures.

A future analysis could extend this work by:

- analysing cancellations and returns;
- performing RFM customer segmentation;
- adding profit or product cost data;
- measuring customer cohorts and retention;
- forecasting demand;
- testing promotional strategies through controlled experiments.
